# FORESIGHT
## Data Loading & Validation

This notebook loads and validates the four core FORESIGHT datasets:

1. Product Master
2. Sales
3. Inventory
4. Calendar

Objectives:
- Loads all datastes
- Inspect dimensions and columns    
- Check data types
- Check missing values
- Check duplicates
- Validate data ranges
- Validate SKU relationships
- Validate sales and inventory coverage

# Import Libraries

```python
import pandas as pd
import numpy as np 
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 120)

print("Libraries imported successfully")
```

# Define Project Root

```python
# Project root
PROJECT_ROOT = Path.cwd().parent

# Data folders
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_PATH = PROJECT_ROOT / "data" / "processed"

print("Project Root :", PROJECT_ROOT)
print("Raw Data     :", RAW_DATA_PATH)
print("Processed    :", PROCESSED_DATA_PATH)
```

# Check Raw Data Folder

```python
if not RAW_DATA_PATH.exists():
    raise FileNotFoundError(f"Raw data folder not found at {RAW_DATA_PATH}. Please ensure the raw data is placed in the correct directory.")
print("Raw data folder exists.\n")

print("Files Available:")
for file in RAW_DATA_PATH.iterdir():
    print(" -",file.name)
```

# Define Datasets Files

```python
PRODUCT_FILE = RAW_DATA_PATH / "product_features.csv"
SALES_FILE = RAW_DATA_PATH / "sales_data.csv"
INVENTORY_FILE = RAW_DATA_PATH / "inventory_data.csv"
CALENDAR_FILE = RAW_DATA_PATH / "calendar_data.csv"

files = {
    "Product Features": PRODUCT_FILE,
    "Sales Data": SALES_FILE,
    "Inventory Data": INVENTORY_FILE,
    "Calendar Data": CALENDAR_FILE
}

for name, path in files.items():
    print(f"{name:15} -> {path}")
```

# Verify All files exists

```python
# Cell 6: Validate required datasets

from pathlib import Path

# Project root
PROJECT_ROOT = Path.cwd().parent

# Raw data directory
DATA_DIR = PROJECT_ROOT / "Data" / "raw"

# Required dataset files
files = {
    "Sales": DATA_DIR / "sales.csv",
    "Inventory": DATA_DIR / "inventory.csv",
    "Products": DATA_DIR / "product_master.csv",
    "Calendar": DATA_DIR / "calendar.csv"
}

# Check for missing files
missing_files = []

for name, path in files.items():
    if not path.exists():
        missing_files.append(f"{name}: {path}")

# Validate files
if missing_files:
    print("❌ Missing files:")
    
    for file in missing_files:
        print(" -", file)
    
    raise FileNotFoundError(
        "One or more required datasets are missing."
    )

print("✅ All four required datasets are present.")

# Display paths
print("\nDataset paths:")
for name, path in files.items():
    print(f" - {name}: {path}")
```

# Load All Datasets

```python
import pandas as pd

#Load datastets
sales_df = pd.read_csv(files["Sales"])
inventory_df = pd.read_csv(files["Inventory"])
products_df = pd.read_csv(files["Products"])
calendar_df = pd.read_csv(files["Calendar"])

print("Datasets loaded successfully.")

# Display datasets shapes
print("\nDataset shapes:")
print(f" - Sales Data: {sales_df.shape}")
print(f" - Inventory Data: {inventory_df.shape}")
print(f" - Products Data: {products_df.shape}")
print(f" - Calendar Data: {calendar_df.shape}")
```

# Check Datasets Columns

```python
datasets = {
    "Sales": sales_df,
    "Inventory": inventory_df,
    "Products": products_df,
    "Calendar": calendar_df
}

for name, df in datasets.items():
    print("\n" + "=" * 60 )
    print(f"{name} Dataset")
    print("*" * 60)

    print("Columns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())
```

# Basic Data Quality check

```python
for name, df in datasets.items():
    print("\n" + "=" * 70)
    print(f"{name} Dataset")
    print("=" * 70)

    # Shape
    print(f"\nRows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")

    # Data types
    print("\nData Types:")
    print(df.dtypes)

    # Missing values
    print("\nMissing Values:")
    missing = df.isnull().sum()
    
    if missing.sum() == 0:
        print("✅ No missing values")
    else:
        print(missing[missing > 0])

    # Duplicate rows
    duplicates = df.duplicated().sum()
    print(f"\nDuplicate Rows: {duplicates:,}")
```

# Converts Date columns

```python

sales_df["Date"] = pd.to_datetime(sales_df["Date"], errors="coerce")
inventory_df["Snapshot_Date"] = pd.to_datetime(
    inventory_df["Snapshot_Date"], 
    errors="coerce"
)
products_df["Launch_Date"] = pd.to_datetime(
    products_df["Launch_Date"], 
    errors="coerce"
)
calendar_df["date"] = pd.to_datetime(
    calendar_df["date"], 
    errors="coerce"
)

print("✅ Date columns converted successfully.")

print("\nDate ranges:")

print(
    "Sales:",
    sales_df["Date"].min(),
    "→",
    sales_df["Date"].max()
)

print(
    "Inventory:",
    inventory_df["Snapshot_Date"].min(),
    "→",
    inventory_df["Snapshot_Date"].max()
)

print(
    "Products:",
    products_df["Launch_Date"].min(),
    "→",
    products_df["Launch_Date"].max()
)

print(
    "Calendar:",
    calendar_df["date"].min(),
    "→",
    calendar_df["date"].max()
)
```

# Check SKU Relationships

```python
sales_skus = set(sales_df["SKU"].dropna().unique())
inventory_skus = set(inventory_df["SKU"].dropna().unique())
products_skus = set(products_df["SKU"].dropna().unique())

print("SKU Counts")
print("-" * 40)

print(f"Sales SKUs     : {len(sales_skus):,}")
print(f"Inventory SKUs : {len(inventory_skus):,}")
print(f"Product SKUs   : {len(products_skus):,}")

print("\nSKU Consistency")
print("-" * 40)

print(
    "Sales → Products:",
    len(sales_skus - products_skus),
    "SKUs missing in Products"
)

print(
    "Sales → Inventory:",
    len(sales_skus - inventory_skus),
    "SKUs missing in Inventory"
)

print(
    "Products → Sales:",
    len(products_skus - sales_skus),
    "SKUs with no sales"
)
```
